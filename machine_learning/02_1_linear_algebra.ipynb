# Linear Algebra: Foundations and Applications

Linear Algebra forms the backbone of modern computational sciences, including machine learning, bioinformatics, and computer graphics. It enables efficient manipulation of large datasets and the mathematical modeling of systems. This blog explores the fundamental concepts of linear algebra, supplemented by detailed examples.

## Why Learn Linear Algebra?

Linear Algebra simplifies:

-   **Data Representation**: Representing high-dimensional datasets as matrices or vectors.
-   **Transformations**: Applying operations like rotation, scaling, and projection.
-   **Optimization**: Solving systems of equations and minimizing error in machine learning models.


## Core Concepts

### Vectors

A **vector** is a quantity with both magnitude and direction. It can represent points in a space, rows in a dataset, or features in machine learning.

#### Example: Representing Data as Vectors

```python
import numpy as np

# Example: 2D and 3D vectors
vector_2D = np.array([3, 4])
vector_3D = np.array([2, -1, 5])

# Magnitude of a vector
magnitude_2D = np.linalg.norm(vector_2D)
print("Magnitude of 2D vector:", magnitude_2D)
```

### Matrices

A **matrix** is a collection of numbers arranged in rows and columns. It is used to store and manipulate multiple vectors or datasets.

#### Example: Matrix Multiplication

```python
#| code-overflow: wrap

# Defining two matrices
A = np.array([[1, 2], [3, 4]])
B = np.array([[2, 0], [1, 3]])

# Matrix multiplication
result = np.dot(A, B)
print("Matrix A:")
print(A)
print("Matrix B:")
print(B)
print("A * B:")
print(result)
```

### Systems of Linear Equations

Systems of equations can be represented as matrix equations $Ax = b$, where $A$ is a matrix, $x$ is a vector of variables, and $b$ is the result.

#### Example: Solving $Ax = b$

```python
# Solving a system of linear equations
A = np.array([[2, 3], [1, 2]])
b = np.array([5, 4])

# Solving for x
x = np.linalg.solve(A, b)
print("Solution for Ax = b:", x)
```

### Eigenvalues and Eigenvectors

Eigenvalues and eigenvectors are critical in understanding matrix transformations, such as scaling and rotation.

#### Example: Finding Eigenvalues and Eigenvectors

```python
# Define a matrix
A = np.array([[2, -1], [-1, 2]])

# Compute eigenvalues and eigenvectors
eigenvalues, eigenvectors = np.linalg.eig(A)
print("Eigenvalues:", eigenvalues)
print("Eigenvectors:")
print(eigenvectors)
```

### Principal Component Analysis (PCA)

PCA is a dimensionality reduction technique using eigenvectors to project data onto lower-dimensional spaces.

$$ x = \sqrt{y} $$

$$ z = \sqrt{a^2 + b^2} $$

$$ \widetilde{\mathbf{A}}_{\Phi} = \mathbf{T}^{-1} \mathbf{A}_{\Phi} \mathbf{S}. $$

$$ \widetilde{b}_j = s_{1j} b_1 + \cdots + s_{nj} b_n = \sum_{i=1}^n s_{ij} b_i, \quad j = 1, \ldots, n. $$

#### Example: Using PCA for Dimensionality Reduction

```python
#| code-fold: true
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

# Create example data
data = np.random.rand(100, 3)

# Apply PCA to reduce dimensions to 2
pca = PCA(n_components=2)
reduced_data = pca.fit_transform(data)

# Plot reduced data
plt.scatter(reduced_data[:, 0], reduced_data[:, 1])
plt.title("2D Projection of 3D Data Using PCA")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.show()
```
