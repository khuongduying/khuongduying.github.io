```python
import plotly.graph_objects as go
import numpy as np

# Function to create the interactive figure
def create_figure():
    # Initial vector components
    x1_init, y1_init = 3, 4
    x2_init, y2_init = 5, -2

    fig = go.Figure()

    # Add traces for the initial vectors
    fig.add_trace(go.Scatter(
        x=[0, x1_init],
        y=[0, y1_init],
        mode='lines+markers',
        name='Vector x',
        line=dict(color='red', width=3)
    ))

    fig.add_trace(go.Scatter(
        x=[0, x2_init],
        y=[0, y2_init],
        mode='lines+markers',
        name='Vector y',
        line=dict(color='blue', width=3)
    ))

    fig.add_trace(go.Scatter(
        x=[0, x1_init + x2_init],
        y=[0, y1_init + y2_init],
        mode='lines+markers',
        name='x + y',
        line=dict(color='green', width=3)
    ))

    # Add sliders to adjust vector components
    steps = []
    for x1 in np.linspace(-10, 10, 21):
        for y1 in np.linspace(-10, 10, 21):
            step = dict(
                method="update",
                args=[{
                    "x": [
                        [0, x1],
                        [0, x2_init],
                        [0, x1 + x2_init]
                    ],
                    "y": [
                        [0, y1],
                        [0, y2_init],
                        [0, y1 + y2_init]
                    ]
                }],
                label=f"x1={x1}, y1={y1}"
            )
            steps.append(step)

    sliders = [dict(
        active=10,
        currentvalue={"prefix": "Vector x (x1, y1): "},
        steps=steps
    )]

    fig.update_layout(
        sliders=sliders,
        title="Interactive Vector Addition",
        xaxis=dict(range=[-10, 10], zeroline=True),
        yaxis=dict(range=[-10, 10], zeroline=True),
        width=800,
        height=600
    )

    fig.show()

# Call the function
create_figure()
```
