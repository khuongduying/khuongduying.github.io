```python
#| echo: false
import matplotlib.pyplot as plt
import numpy as np

# Define vectors
x = np.array([1, 2])
y = np.array([2, 1])

# Resultant vector
resultant = x + y

# Plot
fig, ax = plt.subplots()
ax.quiver(0, 0, x[0], x[1], angles='xy', scale_units='xy', scale=1, color='black', label=r'$\vec{x}$')
ax.quiver(0, 0, y[0], y[1], angles='xy', scale_units='xy', scale=1, color='black', label=r'$\vec{y}$')
ax.quiver(0, 0, resultant[0], resultant[1], angles='xy', scale_units='xy', scale=1, color='black', linestyle='--', label=r'$\vec{x} + \vec{y}$')

# Labels and appearance
ax.text(x[0] / 2, x[1] / 2, r'$\vec{x}$', fontsize=12, color='black', ha='right')
ax.text(y[0] / 2 + 0.2, y[1] / 2, r'$\vec{y}$', fontsize=12, color='black', ha='left')
ax.text(resultant[0] / 2, resultant[1] / 2 + 0.2, r'$\vec{x} + \vec{y}$', fontsize=12, color='black', ha='center')

# Set limits and grid
ax.set_xlim(0, 4)
ax.set_ylim(0, 4)
ax.grid(True, linestyle='--', alpha=0.5)
ax.set_aspect('equal', adjustable='box')

# Title
plt.title("Geometric vectors.", fontsize=14)
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

plt.show()
```
